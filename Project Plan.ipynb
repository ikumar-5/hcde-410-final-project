# A5 Final Project Plan

# 1. Overview & Research Questions

## 1.1 Overview

This project examines whether consumer complaints about digital wallet products look different, and are handled differently, depending on who operates the wallet. Using CFPB complaint data from January to June 2026, I aim to compare three types of providers within the same product category: (1) ecosystem (big-tech) wallets, (2) wallet-first fintech apps, and (3) bank-affiliated services. Digital wallets are a useful setting because ownership and platform control in this market are the subject of active antitrust debate, so differences in how consumers experience these providers are relevant to that conversation. The aim is to characterize what consumers report and how companies respond, not to establish causes or to judge whether any company acted improperly.

## 1.2 Core research questions

**RQ1 (Issue type).** Within digital wallet products, does the mix of issues consumers complain about differ across the three provider types?

**RQ2 (Resolution).** Within digital wallet products, does the way companies respond to complaints (for example, closed with explanation, closed with monetary relief, closed with non-monetary relief) differ across the three provider types?

**Possible extension (only if time and resources permit).**
**RQ3 (Timeliness).** Does the share of complaints answered on time differ across the three provider types?

## 1.3 Provider groups (defined formally in Section 3)

- **G1, Ecosystem wallets:** the wallet is a feature of a large consumer technology platform whose core business is not payments.
- **G2, Wallet-first fintech:** the operator's core business is a payments or fintech app.
- **G3, Bank-affiliated services:** the service is owned or operated by banks or a bank consortium.

## 1.4 Hypotheses

**Exploratory hypotheses (no direction predicted)**
- **H1a (issue type).** The distribution of issues differs across G1, G2, and G3. *Null:* the issue mix is the same across groups.
- **H2 (resolution).** The distribution of company responses differs across G1, G2, and G3. *Null:* the distribution of responses is the same across groups. I do not predict a direction because I found no strong grounds for one in prior work.
- **H3 (extension only).** The share of timely responses differs across groups. *Null:* it is the same.

**Directional hypothesis (grounded in prior reports, not yet established)**
- **H1b (issue type).** Within digital wallets, complaints about wallet-first fintech apps (G2) are relatively more concentrated in account problems (opening, closing, managing, or restricted accounts), while complaints about bank-affiliated services (G3) are relatively more concentrated in fraud and scam-related problems. This expectation rests on prior reporting of frequent account-access complaints about wallet apps and on fraud allegations about a bank-affiliated service (see Section 2). It is directional only for these two issue categories in these two groups.

## 1.5 Scope and what this study will not do

- **Data:** CFPB Consumer Complaint Database, January to June 2026, restricted to the digital wallet sub-product, and to the top N companies by complaint volume (N of roughly 5 to 8, fixed in the first analysis step).
- **Descriptive only:** complaints are self-reported and unverified, so results describe complainants and what they reported, not all customers and not actual harm.
- **No causal claims:** differences across groups may reflect differences in customers, products, or complaint behavior, not company practices.
- **Rule fixed in advance:** the provider-group classification will be written down and applied before any issue or response results are examined.

# 2. Background / Related Work

## 2.1 What is known about digital wallet complaints

Digital wallet complaints, a category that includes peer-to-peer payment apps, have been growing. A 2021 U.S. PIRG Education Fund report counted over 9,000 complaints in this category since the CFPB started accepting them in 2017, with a sharp rise in 2020 and 2021 (U.S. PIRG Education Fund, 2021). It found three frequent issue types: problems opening, closing, or managing accounts; fraud or scams; and transaction problems, including unauthorized transfers. Press coverage of the report also noted that complaint volume was concentrated in a few companies: PayPal (which owns Venmo) and Square (which operates Cash App) had far more complaints than Early Warning Services, the operator of Zelle (Checkbook, n.d.). The report's authors also pointed out that the CFPB database lists these complaints by parent company, not by payment platform, which makes it hard to separate, for example, Venmo from PayPal.

**How this shapes my study:** it supports using account problems and fraud/scams as the issue categories to watch (H1b); it warns that the bank-affiliated group (G3) may have few complaints, which is why my plan includes a feasibility check and a fallback; and it means I must classify companies at the level the data reports them, the parent company.

## 2.2 Platform power, wallets, and antitrust

Economic theory treats payment systems as two-sided markets, where a platform must attract both consumers and merchants and chooses how to price each side (Rochet & Tirole, 2003). Digital wallets have also become part of antitrust enforcement. In March 2024, the U.S. Department of Justice sued Apple, and one of its allegations is that Apple blocks third-party developers from offering tap-to-pay wallets on the iPhone, giving it control over how iPhone users pay (NBC News, 2024; TechCrunch, 2024). The coverage notes that European regulators raised a similar concern about Apple's control of the iPhone's contactless payment technology.

**How this shapes my study:** these sources motivate separating wallets by ownership type, especially ecosystem wallets versus other providers. They concern competition and access, not complaint patterns, so I use them as motivation for the comparison, not as evidence about what complaints should look like.

## 2.3 Bank-affiliated wallets and fraud

Zelle is operated by Early Warning Services, which is owned by seven U.S. banks, and the largest three banks on the network accounted for most of its activity in 2023 (Colorado Biz/AP, 2025). In December 2024 the CFPB sued Early Warning Services and three of those banks, alleging they failed to protect consumers from fraud on Zelle. The agency dismissed the case with prejudice in March 2025 (U.S. District Court for the District of Arizona, 2025). The allegations were never adjudicated, so I treat them only as context for why fraud is a plausible focus for bank-affiliated services.

**How this shapes my study:** it grounds the fraud/scam half of H1b. The 2025 shift in federal enforcement is also a reminder that the regulatory environment around these companies changed shortly before my study period.

## 2.4 What complaint data can and cannot show

Complaint data is not a random sample, so it may not represent all consumers' experiences (Littwin, 2015). The CFPB itself, when it announced in August 2026 that it would stop publishing complaint narratives, described them as unverified and as a less-than-representative sample of consumer experiences (Mondaq, 2026a). Coverage of the database also notes that a very large share of its complaints were filed since January 2025, which is unusual and worth checking in my own data (Mondaq, 2026b).

**How this shapes my study:** it is the reason I frame results as descriptive, avoid claims about harm or company wrongdoing, and plan to check the volume and composition of my sample before analysis.

## 2.5 Methods background

Comparing the distribution of a categorical outcome (issue, response) across groups is a standard use of the chi-square test of independence on a contingency table. Because complaint counts can be large, even small differences become statistically significant, so I will pair the test with an effect size, Cramér's V, both available in SciPy (SciPy documentation, n.d.).

## 2.6 Gap this study addresses

In my search, I found descriptive reporting on digital wallet complaints overall, but I did not find an analysis that compares ownership types within digital wallets by issue mix and complaint outcome. This study is a small, descriptive step in that direction, not a definitive answer.

# 3. Methodology

The analysis for this project will be done in Jupyter notebooks using Python (pandas for data handling, SciPy and statsmodels for statistics, matplotlib/seaborn for figures), so the full pipeline is reproducible in one place. The large CSV files are not committed to the repository; the notebook documents file names and download links.

## 3.1 Data preparation
- Load the January-February (combined), March, April, May, and June 2026 files from the CFPB archive, reading only the columns needed (`usecols`) and using categorical types to limit memory.
- Keep complaints with a date received from January through June 2026, restricted to the digital wallet sub-product (label to be confirmed against the actual column values).

## 3.2 Feasibility check and choosing N (first analysis step)
- Count complaints per company and per provider group.
- Choose the top N companies (about 5 to 8) by total digital wallet complaint volume over the study period. This depends only on volume, not on issue or response results, so it does not bias the findings.
- Set a minimum complaint count per group, before looking at outcomes, below which a group is considered too small for reliable comparison.
- **Fallback:** if a group falls below the minimum, merge it with the closest group or add a second product category, and document the change.

## 3.3 Classifying companies into groups (fixed before looking at outcomes)
Each company is assigned to a group by the core business of its parent company (the level at which the CFPB reports): G1 if the parent is a consumer technology platform whose core business is not payments, G2 if the parent's core business is a payments or fintech app, G3 if the service is owned or operated by banks or a bank consortium. I will record each company, its parent, its group, and the reason in a table, and note ambiguous cases as judgment calls. I will check in the data how complaints about bank-affiliated services are filed (against the operator or against individual banks) and describe how I handle this.

## 3.4 Analysis, matched to each question
- **RQ1 (issue type):** cross-tabulate issue by group and show row-normalized shares. Test with a chi-square test of independence (`scipy.stats.chi2_contingency`) and report Cramér's V (`scipy.stats.contingency.association`) as the effect size. Rare issues are collapsed into "Other" using a threshold set in advance so expected cell counts stay adequate. Report confidence intervals for key shares (e.g., with statsmodels).
- **RQ2 (resolution):** the same approach for the company response categories. I will also collapse responses into a simpler "any relief vs. no relief" outcome and compare proportions across groups, since the detailed categories may be applied inconsistently.
- **Extension, RQ3 (timeliness):** compare the share of timely responses across groups using the same tools.

**Why these methods fit:** the outcomes (issue, response, timely or not) are categorical, and the question is whether their distribution differs across groups, which is what contingency-table tests address. Because the sample may be large, I rely on effect sizes and intervals, not p-values alone. The design is descriptive, so I do not use causal models.

## 3.5 Robustness checks
- **By company:** repeat the comparison within each group by company, to see whether a single company drives a group's result.
- **By month:** compare patterns across January to June to see whether they are stable. I will note that the CFPB changed aspects of its complaint process in 2026, which could affect month-to-month comparisons.

## 3.6 Presenting findings
Findings will be presented as figures followed by short written descriptions: grouped or stacked bar charts of issue and response shares by group with confidence intervals, small multiples by month, company-level breakdowns, and a summary table of test results and effect sizes. Each figure will be followed by a brief plain-language explanation of what it shows and what it does not show.

## 3.7 Limitations
Complaints are self-reported, unverified, and not representative of all customers; group assignment is my own judgment; companies are reported by parent; and the study is descriptive, so differences cannot be attributed to company practices.
