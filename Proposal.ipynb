# HCDE 410 Final Project Proposal

## Complaints About Digital Financial Products: Platform-Style Fintech vs. Traditional Providers

## 1. Motivation & Problem

Digital platforms increasingly mediate how consumers handle money, and when something goes wrong (a frozen account, an unauthorized transfer, a confusing fee), complaints are one of the few public traces of that friction. The aime of this project is to use the CFPB Consumer Complaint Database to ask:

> **How do consumer complaints about digital financial products differ between platform-style fintech companies and traditional providers, in volume (relative to company size), issue type, and resolution outcome, from January to July 2026?**

I am interested in this for three reasons. First, it sits at the intersection of platform economics and antitrust and consumer protection policy, areas I follow closely. Second, it connects to a research project I am working on which explores how visual salience affects equilibrium selection in Stag Hunt coordination games. Interface design shapes the choices people make, and complaint data offers a real-world, observational counterpart to what I study in a controlled setting. Third, I want to practice turning messy administrative data into evidence a policymaker could use.

I hope to learn whether platform-style products show systematically different complaint patterns, and what a descriptive analysis of self-reported complaints can and cannot support. Because complaints are self-selected and unverified, I will treat results as descriptive, not as measures of actual harm or company wrongdoing.

**Sub-questions**
1. **Volume:** Do platform-style companies receive more or fewer complaints than traditional providers once company size is accounted for?
2. **Issue type:** What are the complaints about (account freezes, unauthorized transactions, fees, support access), and does the mix differ between the two groups?
3. **Resolution:** How do companies respond (explanation only, monetary relief, non-monetary relief), and does that differ by group?

## 2. Data

**Dataset:** CFPB Consumer Complaint Database Narratives Archive, published in the CFPB's FOIA electronic reading room.

**Link:** https://www.consumerfinance.gov/foia-requests/foia-electronic-reading-room/cfpb-consumer-complaint-database-narratives-archive/

**Files used:** The CSV exports for January-February 2026 (combined), March, April, May, June, and July 2026 (about 250 MB each). July 2026 is the primary month, earlier months serve as a check on whether patterns hold.

**What it contains:** One row per complaint submitted to the CFPB about a financial product or service. Fields include the product and sub-product, issue and sub-issue, company, state, how the complaint was submitted, the company's response, and whether the response was timely. Many rows also include the consumer's narrative, published with the consumer's consent. *(Exact column names to be confirmed once the files are loaded.)*

**Why it fits:** It records real consumer interactions with digital financial products at scale, with enough structure (product, issue, company, outcome) for grouped statistical comparisons and enough free text to classify issue types. The monthly files also allow a simple check across time.

**License/terms:** The archive page does not state a specific license. The data is published by a U.S. federal agency through its public FOIA reading room. I will cite the CFPB as the source, link to the archive page, and follow the terms in the CFPB's website legal notices. The large CSV files will not be redistributed in the repo (see below).

**Ethical considerations:**
- Complaints are self-selected and unverified, so they cannot establish actual harm or wrongdoing by any company.
- Narratives are published only with consent and are scrubbed, but they may still contain personal details. I will not publish individual narratives or try to identify anyone.
- Naming specific companies has reputational stakes, so results will be framed descriptively and with caveats.
- People who file complaints are not a representative sample of all customers, so findings describe complainants, not all users.

**Reproducibility:** The CSVs are too large for GitHub (100 MB file limit), so they are excluded via `.gitignore`. The notebook will document the file names and download links so anyone can rerun the analysis.

## 3. Analysis Methods

I plan to use **Python (pandas, SciPy, statsmodels, matplotlib/seaborn)** in Jupyter notebooks as the main tool, so the whole analysis is reproducible in one place. Planned steps:
- Clean and combine the monthly files, filter to digital-product categories, and classify companies as platform-style or traditional using a documented rule.
- Compare complaint shares, issue mix, and response outcomes between groups using descriptive statistics and tests of proportions (e.g., chi-square).
- Fit a simple regression (e.g., logistic regression on whether a complaint received relief, controlling for product, state, and submission channel).
- Classify issue types in the narrative text with a simple keyword or TF-IDF approach.
- Create visualizations in Python, and optionally build an exploratory dashboard in Tableau if I have access.

I may use R to cross-check the statistical results, but Python will be the primary, reproducible pipeline.

## 4. Unknowns & Dependencies

- **Source stability:** The CFPB stopped publishing complaint narratives in the live database in August 2026 and now offers them only through this archive. The archive page could change or move, so I will download and keep local copies and record the retrieval date.
- **Policy changes during the study period:** The CFPB changed its complaint intake process during 2026, including a February notice discouraging certain credit reporting complaints and further changes announced in June. Month-to-month differences could reflect these changes, not company behavior. I will focus on digital-product categories and note this limitation.
- **Data volume and quality:** Complaint volume has surged since early 2025, and closure categories like "Closed with non-monetary relief" have been inconsistently applied. I will check the data for anomalies before drawing conclusions.
- **Platform vs. traditional classification:** This is my own judgment, so results may depend on where I draw the line. I will document the rule and test sensitivity to it.
- **Company size data:** Normalizing complaints by company size requires outside data (e.g., customer counts) that may be hard to find. *Fallback:* compare shares of complaints within product categories instead.
- **Compute limits:** About 1.5 GB of CSV may strain JupyterHub memory. I will load only needed columns, use categorical types, and save a slimmed Parquet file.
- **Tools and time:** Tableau access is not guaranteed, and the term schedule limits scope. *Fallback:* use Python-only visualizations and July data only.
