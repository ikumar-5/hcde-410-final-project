## 1. Motivation & Problem

Digital platforms increasingly mediate how consumers handle money, and when something goes wrong (a frozen account, an unauthorized transfer, a confusing fee), complaints are one of the few public traces of that friction. The aim of this project is to use the CFPB Consumer Complaint Database to ask:

> **How do consumer complaints about digital wallet products differ across ecosystem (big-tech), wallet-first fintech, and bank-affiliated providers, in issue type and in how companies respond, from January to June 2026?**

I am interested in this for three reasons. First, it sits at the intersection of platform economics and antitrust and consumer protection policy, areas I follow closely, and digital wallets are a market where platform control is the subject of active antitrust debate. Second, it connects to a research project I am working on which explores how visual salience affects equilibrium selection in Stag Hunt coordination games. Interface design shapes the choices people make, and complaint data offers a real-world, observational counterpart to what I study in a controlled setting. Third, I want to practice turning messy administrative data into evidence a policymaker could use.

I hope to learn whether complaints about different types of wallet providers show systematically different patterns, and what a descriptive analysis of self-reported complaints can and cannot support. Because complaints are self-selected and unverified, I will treat results as descriptive, not as measures of actual harm or company wrongdoing.

**Core questions**
1. **Issue type:** What are the complaints about (for example account problems, fraud and scams, transaction problems), and does the mix differ across the three provider types?
2. **Resolution:** How do companies respond (explanation only, monetary relief, non-monetary relief), and does that differ across the three provider types?

**Possible extension (if time and resources permit)**
3. **Timeliness:** Does the share of complaints answered on time differ across the three provider types?

## 2. Data

**Dataset:** CFPB Consumer Complaint Database Narratives Archive, published in the CFPB's FOIA electronic reading room.

**Link:** https://www.consumerfinance.gov/foia-requests/foia-electronic-reading-room/cfpb-consumer-complaint-database-narratives-archive/

**Files used:** The CSV exports for January-February 2026 (combined), March, April, May, and June 2026 (about 250 MB each). I will restrict the analysis to the digital wallet sub-product and to the top N companies (about 5 to 8) by complaint volume, using complaints received from January through June 2026.

**What it contains:** One row per complaint submitted to the CFPB about a financial product or service. Fields include the product and sub-product, issue and sub-issue, company, state, how the complaint was submitted, the company's response, and whether the response was timely. Many rows also include the consumer's narrative, published with the consumer's consent, though the narratives are not part of my core analysis. *(Exact column names to be confirmed once the files are loaded.)*

**Why it fits:** It records real consumer interactions with digital wallet products at scale, with enough structure (product, issue, company, response) for grouped statistical comparisons. The CFPB reports complaints by parent company, which matches how I will classify providers into groups. The monthly files also allow a simple check across time.

**License/terms:** The archive page does not state a specific license. The data is published by a U.S. federal agency through its public FOIA reading room. I will cite the CFPB as the source, link to the archive page, and follow the terms in the CFPB's website legal notices. The large CSV files will not be redistributed in the repo (see below).

**Ethical considerations:**
- Complaints are self-selected and unverified, so they cannot establish actual harm or wrongdoing by any company.
- Narratives are published only with consent and are scrubbed, but they may still contain personal details. I will not publish individual narratives or try to identify anyone.
- Naming specific companies has reputational stakes, so results will be framed descriptively and with caveats.
- People who file complaints are not a representative sample of all customers, so findings describe complainants, not all users.

**Reproducibility:** The CSVs are too large for GitHub (100 MB file limit), so they are excluded via `.gitignore`. The notebook will document the file names and download links so anyone can rerun the analysis.

## 3. Analysis Methods

I plan to use **Python (pandas, SciPy, statsmodels, matplotlib/seaborn)** in Jupyter notebooks, so the whole analysis is reproducible in one place. Planned steps:
- Combine the monthly files, keeping only the columns I need, and filter to the digital wallet sub-product.
- Choose the top N companies by complaint volume and classify each into one of three provider groups (ecosystem, wallet-first fintech, bank-affiliated) using a documented rule written before I look at any results.
- Compare issue mix and response outcomes across groups using contingency tables, a chi-square test of independence, and Cramér's V as an effect size.
- Check robustness by company and by month.
- Present findings as visualizations in Python, each followed by a brief explanation.

The detailed methodology is in *Project Plan.ipynb*.

## 4. Unknowns & Dependencies

- **Source stability:** The CFPB stopped publishing complaint narratives in the live database in August 2026 and now offers them only through this archive. The archive page could change or move, so I will download and keep local copies and record the retrieval date.
- **Policy changes during the study period:** The CFPB changed its complaint intake process during 2026, including a February notice discouraging certain credit reporting complaints and further changes announced in June. Month-to-month differences could reflect these changes, not company behavior. I will focus on digital wallet complaints and note this limitation.
- **Data volume and quality:** Complaint volume has surged since early 2025, and closure categories like "Closed with non-monetary relief" have been inconsistently applied. I will check the data for anomalies before drawing conclusions.
- **Provider classification:** Assigning companies to the three groups is my own judgment, so results may depend on where I draw the line. I will document the rule, apply it before looking at results, and note ambiguous cases.
- **Small groups:** One or more groups, especially the bank-affiliated group, may have too few complaints for reliable comparison. *Fallback:* merge the group with the closest one or add a second product category, and document the change. Complaints about bank-affiliated services may also be filed under individual banks, so I will check how they appear in the data.
- **Compute limits:** Over a gigabyte of CSV may strain JupyterHub memory. I will load only needed columns, use categorical types, and save a slimmed Parquet file.
- **Time:** The term schedule limits scope. *Fallback:* drop the timeliness extension first, then reduce the number of months.
